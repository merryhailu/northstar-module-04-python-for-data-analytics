## Artifact 1: A cleaned, enriched customer DataFrame.

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

df = pd.read_csv('finflow-customers-200k.csv',
                parse_dates = ['signup_date', 'last_login_date', 'churn_date'])

print(f'Customer Data: {df.shape}')
```

## load plan metadata

```python
plans_raw = pd.read_csv('plan_metadata_raw.csv',
                       parse_dates= ['launch_date'])
print(f'Plans Data : {plans_raw.shape}')
plans_raw
```

```python
#survey the problems
print('Unique plan tier values')
for value in plans_raw['plan_tier'].unique():
    print(f" '{value}' : ({len(value)})")
print()

#missing Values
print('Missing values per columns')
print(plans_raw.isna().sum())
print()

#Duplicates
print('Duplication')
print(plans_raw.duplicated().sum())
```

**Observation:** Five rows total, four distinct after deduplication.
Plan tier values have whitespace and capitalization mess that will prevent
the join column from matching the customer data. monthly_fee_listed is
missing for Legacy. Need to clean before merging.

```python
# Strip whitespace, then title-case
plans_raw['plan_tier_clean'] = plans_raw['plan_tier'].str.strip().str.title()

print('Cleaned plan_tier')
for value in plans_raw['plan_tier_clean'].unique():
    print(f"{value}: (length {len(value)})")
```

```python
# How many rows before
print(f"Rows before dedupe: {len(plans_raw)}")

# Drop duplicates on the cleaned column
plans = plans_raw.drop_duplicates(subset=['plan_tier_clean'], keep='first').copy()

print(f"Rows after dedupe: {len(plans)}")
plans
```

```python
# Confirm the NaN remains
print('Monthly fee listed values')
print(f'{plans[['plan_tier_clean', 'monthly_fee_listed']]}')
```

**Observation:** Leaving the Legacy monthly_fee_listed as NaN rather than
filling. The customer data already has a per-customer monthly_fee, so the
listed price isn't critical for the join. If Hannah asks during the
readout, we can investigate.

```python
print("Customer plan_tier values:")
print(df['plan_tier'].value_counts())
```

```python
plans = plans.drop(columns=['plan_tier'])

print("Metadata columns going into the merge:", list(plans.columns))
```

```python
#left join
df_enriched = pd.merge(
    df,
    plans,
    left_on= 'plan_tier',
    right_on='plan_tier_clean',
    how='left'
)
print(f'shape after merge: {df_enriched.shape}')
```

```python
df_enriched.to_csv('finflow-enriched-200k.csv', index=False)
print('Saved to finflow-enriched-200k.csv')
```

## Artifact 2: The consolidated per-tier memo table.

```python
# Target segment
target_segment = 

# customer count 
customer_count = df_enriched.groupby('marketing_name')['customer_id'].count()

# Total mrr from Active
total_mrr = (
    df_enriched[df_enriched['active']]
    .groupby('marketing_name')['monthly_fee']
    .sum())
    
# churn rate by marketing_name
def churn_rate_pct(active_series):
    return (~active_series).mean() * 100

churn_rate = (
    df_enriched.groupby('marketing_name')['active']
    .apply(churn_rate_pct)
)

# Average sessions among active customers 
avg_sessions_active = (
    df_enriched[df_enriched['active']]
    .groupby('marketing_name')['total_sessions']
    .mean()
)
# Average sessions among churned customers
avg_sessions_churned = (
    df_enriched[~df_enriched['active']]
    .groupby('marketing_name')['total_sessions']
    .mean()
)
# mrr pct of total
mrr_pct_of_total = (total_mrr/ total_mrr.sum())*100

# Combine everything into one DataFrame
memo_summary_final = pd.DataFrame({
     'target_segment' : target_segment,
     'customer_count': customer_count,
     'churn_rate_pct': churn_rate,
     'total_mrr': total_mrr,
     'avg_sessions_active': avg_sessions_active,
     'avg_sessions_churned': avg_sessions_churned,
     'mrr_pct_of_total': mrr_pct_of_total
})
   
print(memo_summary_final)
```
