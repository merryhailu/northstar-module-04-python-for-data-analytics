## Re-loading with proper date parsing

`pd.read_csv` accepts a `parse_dates` parameter that converts named columns
from string to datetime during load. Always do this for date columns; many
Pandas operations work differently (and better) on real datetimes.

```python
df = pd.read_csv('finflow-customers-1000.csv',
                parse_dates=['signup_date', 'last_login_date', 'churn_date'])
print(df.dtypes[['signup_date', 'last_login_date', 'churn_date']])
```

## Building a DataFrame from scratch

For test data, reference tables, and ad-hoc lookups, you often build a
small DataFrame inline rather than loading one from a file.

```python
# A small reference table of FinFlow's plan pricing
plans = pd.DataFrame({
    'plan_tier': ['Basic','Pro','Enterprise'],
    'monthly_fee': [9.99, 29.99, 199.99],
    'annual_value': [9.99 *12, 29.99 *12, 199.99*12]
})
plans
```

## Saving a DataFrame to CSV

The plans table we just built doesn't exist as a file anywhere. Save it
so we can load it again later if needed.

```python
plans.to_csv('finflow-plans.csv',index=False)
print("saved finflow-plans.csv")
```
