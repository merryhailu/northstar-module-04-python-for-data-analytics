```python
# Imports for this lesson
# For now, we use only the standard library.

import random
import statistics
```

```python
finflow_sample = [
    {"customer_id": "FF-10001", "plan_tier": "Basic",    "monthly_fee": 9.99,  "months_active": 4,  "active": True},
    {"customer_id": "FF-10002", "plan_tier": "Pro",      "monthly_fee": 29.99, "months_active": 14, "active": True},
    {"customer_id": "FF-10003", "plan_tier": "Basic",    "monthly_fee": 9.99,  "months_active": 2,  "active": False},
    {"customer_id": "FF-10004", "plan_tier": "Pro",      "monthly_fee": 29.99, "months_active": 18, "active": True},
    {"customer_id": "FF-10005", "plan_tier": "Enterprise","monthly_fee": 199.00,"months_active": 24, "active": True},
    {"customer_id": "FF-10006", "plan_tier": "Basic",    "monthly_fee": 9.99,  "months_active": 1,  "active": False},
    {"customer_id": "FF-10007", "plan_tier": "Pro",      "monthly_fee": 29.99, "months_active": 6,  "active": False},
    {"customer_id": "FF-10008", "plan_tier": "Pro",      "monthly_fee": 29.99, "months_active": 22, "active": True},
    {"customer_id": "FF-10009", "plan_tier": "Basic",    "monthly_fee": 9.99,  "months_active": 8,  "active": True},
    {"customer_id": "FF-10010", "plan_tier": "Enterprise","monthly_fee": 199.00,"months_active": 9,  "active": False},
]
```

Analysis 1. Count the total number of active customers and the total number of churned (active == False) customers. Print both. Add a markdown observation noting the churn rate as a percentage.

**Observation:** Of the 10 customers in the FinFlow sample, 6 are active and 4 have churned (`active == False`). This gives a churn rate of 4 / 10 = **40%**, which is high and worth investigating further in the next steps of the analysis.

```python
churned_count = 0
active_count = 0
for customer in finflow_sample:
    if customer['active']:
        active_count += 1
    else:
        churned_count += 1

churn_rate = churned_count / (active_count + churned_count) * 100

print(f'Total number of Active customers: {active_count}')
print(f"Total number of Churned: {churned_count}")
print(f'Churn rate: {churn_rate:.1f}%')
```

Analysis 2. Compute the total monthly revenue from currently active customers. Print it formatted as US dollars. Add a markdown observation noting how Enterprise customers compare to Basic in terms of contribution.

**Observation:** Active customers generate **$308.95** per month in total. Enterprise customers contribute **$199.00 (64.4% of revenue)** from only **1 customer**, while Basic customers contribute **$19.98 (6.5% of revenue)** from **2 customers**. This shows that Enterprise earns far more per customer, even though Basic has more customers. Retaining Enterprise customers matters most for revenue, since losing that one account would remove nearly two-thirds of monthly revenue.


```python
for tier in ['Basic', 'Pro', 'Enterprise']:
    tier_revenue = sum(
        customer['monthly_fee']
        for customer in finflow_sample
        if customer['active'] and customer['plan_tier'] == tier
    )
    print(f'{tier} monthly revenue: ${tier_revenue:,.2f}')
```

Analysis 3. For each plan tier (Basic, Pro, Enterprise), print the count of churned customers in that tier. Use a loop or a dictionary. Add a markdown observation about which tier appears most vulnerable to churn

**Observation:** The **Basic** tier has the highest number of churned customers, with **2** (half of all 4 churned customers), compared with 1 each for Pro and Enterprise. This suggests Basic may be the most vulnerable tier. However, the count alone does not give the churn rate. 

```python
churned_count_basic = 0
churned_count_pro = 0
churned_count_enterprise = 0

for customer in finflow_sample:
    if not customer['active'] and customer['plan_tier'] == 'Basic' :
        churned_count_basic += 1
    if not customer['active'] and customer['plan_tier'] == 'Pro' :
        churned_count_pro += 1
    if not customer['active'] and customer['plan_tier'] == 'Enterprise' :
        churned_count_enterprise += 1
        
print(f'Curned customer in Basic tier: {churned_count_basic}')
print(f'Curned customer in Pro tier: {churned_count_pro}')
print(f'Curned customer in Enterprise tier: {churned_count_enterprise}')
```

Analysis 4. Compute the average months_active for churned customers. Print it. Add a markdown observation about what this number suggests (do customers churn early, mid-tenure, or late?).

**Observation:** Churned customers stayed an average of **4.5 months**, which suggests churn tends to happen early in the customer lifecycle. The individual values (2, 1, 6, and 9 months) show a split: half of the churned customers left within the first 2 months, while the other half lasted 6 to 9 months. With only 4 churned customers, this is a small sample, but it suggests the early months are the highest-risk period.

```python
churned_months_active = [customer['months_active'] for customer in finflow_sample if not customer['active']]

print(f'{churned_months_active}')
print(f'average Months_active for churned customers: {statistics.mean(churned_months_active)}')
```

Analysis 5. Write a function named lifetime_revenue(customer) that takes a customer dictionary and returns monthly_fee * months_active as the customer's lifetime revenue so far. Use the function to compute the total lifetime revenue across all ten customers. Print the total formatted as US dollars. Add a markdown observation about what this represents.

**Observation:** Total lifetime revenue is **$8,516.25**, the sum of each customer's monthly fee multiplied by the months they have been active, across all ten customers (both active and churned). 

```python
def lifetime_revenue(customer):
   return customer['monthly_fee'] * customer['months_active']

total_lifetime_revenue =sum (
    lifetime_revenue(customer) for customer in finflow_sample) 

print(f'Total lifetime revenue: ${total_lifetime_revenue:,.2f}')
```
