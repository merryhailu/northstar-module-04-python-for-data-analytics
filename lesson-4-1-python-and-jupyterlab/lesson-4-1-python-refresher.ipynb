## Lists

Lists hold ordered sequences. A common analyst use is holding a numeric series:
monthly revenue, daily counts, individual order amounts.

```python
# Monthly revenue for a fictional Q4 (October, November, December)
monthly_revenue = [128500, 134200, 198750]

#inspect the list
print("monthly_revenue:", monthly_revenue)
print("number of months:", len(monthly_revenue))
print("First month:", monthly_revenue[0])
print("Last month:", monthly_revenue[-1])

# Slicing: October and November only
print("first two month:", monthly_revenue[0:2])

# Aggregations using the statistics library
print("sum:", sum(monthly_revenue))
print("Mean:", statistics.mean(monthly_revenue))
print("Median:", statistics.median(monthly_revenue))
```

**Observation:**
Mean is materially higher than median because December's
revenue is roughly 50% higher than the previous two months. Holiday-season
seasonality. In Module 1 we'd use median or compute a confidence interval
to handle this. In Pandas (next lesson), we'll do the same operations on
much larger data.

## Dictionaries

Dictionaries hold record-like data: a single row, a configuration object,
a customer profile. Access values by name rather than position.

```python
# A single fictional customer record
customer = {
    "customer_id": "CUST-49281",
    "first_name": "Aiyana",
    "last_name": "Saunders",
    "email": "aiyana.saunders@example.com",
    "lifetime_value": 1247.50,
    "churned": False,
    "first_order_date": "2023-04-12"   
}

# Access by key name
print("Customer ID:", customer['customer_id'])
print("Life time value: $", customer['lifetime_value'])

# Iterate over keys and valuesabs
print("\nAll fields")
for field_name , value in customer.items():
    print(f"{field_name}: {value}" )
```

**Observation:** Dictionaries are how you'll model individual rows when you
need to manipulate them in Python before loading into a DataFrame. In Lesson
4.2 we'll generate thousands of these with the Faker library.

## A list of dictionaries

A list of dictionaries is the most common pre-DataFrame data shape in Python.
Each dictionary is a row. The list is the table. When you load a CSV or call
an API that returns JSON, you usually start with this shape. Then you
convert it to a DataFrame.

```python
# A small list of fictional customers
customers = [
    {"customer_id": "CUST-49281", "lifetime_value": 1247.50, "churned": False},
    {"customer_id": "CUST-49282", "lifetime_value": 89.00, "churned": True},
    {"customer_id": "CUST-49283", "lifetime_value": 3502.10, "churned": False},
    {"customer_id": "CUST-49284", "lifetime_value": 412.75, "churned": True},
    {"customer_id": "CUST-49285", "lifetime_value": 1855.30, "churned": False},
]

# Count churned vs non-churned with a loop

churned_count = 0
retained_count = 0
for c in customers:
    if c['churned']:
        churned_count = churned_count + 1
    else:
        retained_count = retained_count + 1

print(f"churned: {churned_count}")
print(f"Retained: {retained_count}")

# The same count with a list comprehension
churned_count_comp = sum(1 for c in customers if c['churned'])
print(f'churned (via comprehension): {churned_count_comp}')

# Average LTV for churned customers
churned_ltvs = [c['lifetime_value'] for c in customers if c['churned']]
print(f'churned ltvs: {churned_ltvs}')
print(f'average ltvs churned customer: ${statistics.mean(churned_ltvs):.2f}')
```

## A function

Functions wrap reusable logic with a clear name. Working analysts use them
to avoid copying and pasting the same calculation in five places.

```python
def compute_yoy_growth(prior_period, current_period):
    """Return year-over-year growth as a decimal (e.g., 0.15 = +15%).
     Returns None if prior_period is zero (avoiding division by zero).
    """
    if prior_period == 0:
        return None
    return (current_period - prior_period)/prior_period

# Apply the function to three fictional stores

stores = [
    {"store_id": "S-001", "rev_2023": 1200000, "rev_2024": 1380000},
    {"store_id": "S-002", "rev_2023": 850_000,   "rev_2024": 815_000},
    {"store_id": "S-003", "rev_2023": 0,         "rev_2024": 420_000},
]

for store in stores:
    growth = compute_yoy_growth(store['rev_2023'], store['rev_2024'])
    if growth is None:
        print(f'{store['store_id']}: no prior data')
    else:
        print(f'{store['store_id']}: {growth:.1%} yoy')
```
